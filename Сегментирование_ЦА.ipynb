```python
words_1 = ['мам', 'дет', 'доч', 'сын', 'семь', 'муж', 'жен', 'супруг'] #родители с детьми
words_2 = ['работ', 'ноут', 'розет', 'коворк', 'созвон', 'тих', 'интернет'] #фрилансеры
words_3 = ['тих', 'тиш', 'учить', 'учу', 'фокус', 'занимать', 'занимаюсь', 'готов', 'бот'] #студенты
words_4 = ['гов', 'болт',  'покой', 'отдых', 'недалеко', 'близко'] #жители поблизости (пришли ради вайба)
words_5 = ['вкус', 'терп', 'качеств'] #ценители кофе

count_1 = 0
count_2 = 0
count_3 = 0
count_4 = 0
count_5 = 0


def contains_words_1(cell):
    return any(word in str(cell) for word in words_1)
def contains_words_2(cell):
    return any(word in str(cell) for word in words_2)
def contains_words_3(cell):
    return any(word in str(cell) for word in words_3)
def contains_words_4(cell):
    return any(word in str(cell) for word in words_4)
def contains_words_5(cell):
    return any(word in str(cell) for word in words_5)



result_1 = df.applymap(contains_words_1)
count_1 = result_1.sum().sum()

result_2 = df.applymap(contains_words_2)
count_2 = result_2.sum().sum()

result_3 = df.applymap(contains_words_3)
count_3 = result_3.sum().sum()

result_4 = df.applymap(contains_words_4)
count_4 = result_4.sum().sum()

result_5 = df.applymap(contains_words_5)
count_5 = result_5.sum().sum()

print("Количество ячеек, содержащих потенциальные слова указатели на родителей с детьми", count_1)
print("Количество ячеек, содержащих потенциальные слова указатели на фрилансеров", count_2)
print("Количество ячеек, содержащих потенциальные слова указатели на студентов", count_3)
print("Количество ячеек, содержащих потенциальные слова указатели на жителей близких домов", count_4)
print("Количество ячеек, содержащих потенциальные слова указатели на ценителей кофе", count_5)
```
