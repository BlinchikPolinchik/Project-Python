```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://moluch.ru/archive/280/63177/?ysclid=m8j41cs4xh215685554"

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, 'html.parser')

table = soup.find('table')

rows = table.find_all('tr')

column_1 = []
column_2 = []

for row in rows:
    columns = row.find_all('td')
    col1 = columns[0].text.strip()
    col2 = columns[1].text.strip()
    column_1.append(col1)
    column_2.append(col2)

df = pd.DataFrame({
    column_1[0]: column_1[1:],
    column_2[0]: column_2[1:]
})

print(df)


import plotly.express as px

fig = px.pie(df, values='Кол- во человек', names='Регулярность посещения кофеен',
             title='Регулярность посещения кофеен', color_discrete_sequence=px.colors.qualitative.Pastel)

fig.show()
```
